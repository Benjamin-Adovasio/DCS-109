## 1f (a little trickier!)

Write a function called align_strands(strand_list) that takes a list of DNA strands, like: ['ATTACGA', 'GTGTAC', 'ATC']), and which prints them so they're right-aligned, with leading '_' characters used to pad shorter strands. For example, running your function on the strands given above should produce something like:

ATTACGA<br>
__GTGTAC<br>
_____ATC<br>

Sequence alignment is a very fundamental for any analysis with DNA, so it's important to get it right! **Hints:** Figure out the longest string and compute the length of each string. That will tell you how much padding you'll need. Remember that arithmetic operators like + and * have different meanings in "string world". You can use that to your advantage here.


```python
# Your code here:
def align_strands(strand_list):
  max_len = max(len(s) for s in strand_list)
  for strand in strand_list:
    padding = '_' * (max_len - len(strand))
    print(padding + strand)

# Running this code should produce the expected behavior:
align_strands(['ATTACGA', 'GTGTAC', 'ATC'])
```

# 1g

It's sometimes said that "what happens in a function, STAYS in a function." Consider this example below, in which we create a variable x inside the function modify_local(), that seems like it should overwrite the x first defined at the top. In fact, though, it's like these two x's are on different planets and not  affecting each other. The technical way for saying this is that variables inside functions only have **local scope**. They're only visible from within the function, and vanish when the function finishes.

```python
x = 10

def modify_local():
    x = 5
    print(f'inside the function x has the value {x}')

modify_local()
print(f'outside the function x has the value {x}')
```

To make your brain hurt even more, consider this, which seems like it **definitely** should change the value of x. But in fact, it doesnt...

```python
x = 10

def modify_local(x):
    x = 5
    print(f'inside the function x has the value {x}')

modify_local(x)
print(f'outside the function x has the value {x}')
```

Below is the same code, but with annotations explaining what's happening:

```python
x = 10 # This is a *GLOBAL* variable

def modify_local(x): # This is a *LOCAL* variable, visible only to the function
    x = 5 # Here we assign a new value to the local variable
    print(f'inside the function x has the value {x}')

modify_local(x) # We pass the argument '10' to the function, by VALUE, but never do anything to the global variable x.
print(f'outside the function x has the value {x}')
```

## 2b
 Dictionaries are very flexible data structures, allowing you to even make a dictionary where the values are themselves other dictionaries. Consider this:

```python
gradebook = {
    'Alice': {'exam1': 85, 'exam2': 92},
    'Bob': {'exam1': 78, 'exam2': 89},
    'Charlie': {'exam1': 95, 'exam2': 98}
}
```

Here's how you can iterate over all the keys:

```python
for student in gradebook:
  for exam in gradebook[student]: # note that gradebook[student] is an entire dictionary: {'exam1': 85, etc...}
    print(f'{student} got a {gradebook[student][exam]} on {exam}')
```

Below, write a function `max_grade(gradebook, assignment)` that takes a gradebook dictionary and an assignment name as parameters, and which returns the max score on that assignment across **all** students.

**Hints:** 1) A nested for-loop is overkill for this. But if you understand the code above, you should pretty easily be able to do what's needed to solve this problem. 2)You'll want to initialize a variable or list to keep track of a quantity that changes as you loop. Can you think of a clever use of a conditional (i.e. 'if') so that you only update if you encounter a new maximum? 3) As above, try to solve for a specific case before writing a function.


```python
# Your code below:

def max_grade(gradebook, assignment):
  max_score=0
  for student in gradebook:
    if assignment in gradebook[student]:
      score=gradebook[student][assignment]
      if score>max_score:
        max_score=score
  return max_score

# Running this code should produce the correct answer (i.e. 95 in the given case):
max_grade(gradebook, 'exam1')
```

```python
# keys are grocery store items, and values are quantity to purchase:
shopping_list = {
    'eggs': 2,
    'apples': 5,
    'milk': 1,
    'bread': 3
}

# keys are those same grocery store items, and values are prices per (single) item, in $USD:
prices = {
    'apples': 1.99,
    'bread': 2.99,
    'butter': 4.99,
    'eggs': 4.50,
    'milk': 2.49,
    'oranges': 2.99,
    'potatoes': 1.99,
    'yogurt': 3.74
}
```

Can you write a function that takes a shopping list dictionary and a prices dictionary, and which returns the total cost of the shopping list?

```python
# Your code here:
def total_cost(shopping_list, prices):
  total =0
  for item, quantity in shopping_list.items():
    if item in prices:
      total += quantity * prices[item]
  return total
print(total_cost(shopping_list, prices))
```
